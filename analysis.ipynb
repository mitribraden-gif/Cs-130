```python
with open("cars.csv", "r") as file:
    data = file.readlines()
    data_list = []
for line in data:
    line = line.strip()          
    row = line.split(",")        
    data_list.append(row)
for row in data_list[1:]:  # skip header if there is one
    if float(row[2]) > 50:
        print(row)
```
